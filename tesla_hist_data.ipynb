```python
def make_graph(df, df2, company):

    df["Date"] = pd.to_datetime(df["Date"])
    df2["Date"] = pd.to_datetime(df2["Date"])

    df2 = df2.dropna(subset=["Revenue (Billion USD)"])
    df2 = df2.sort_values("Date")

    fig = make_subplots(
        rows=2, cols=1,
        subplot_titles=[
            f"{company} Historical Stock Price",
            f"{company} Annual Revenue"
        ],
        vertical_spacing=0.15
    )

    fig.add_trace(
        go.Scatter(
            x=df["Date"], y=df["Close"],
            mode="lines", name="Stock Price"
        ),
        row=1, col=1
    )

    fig.add_trace(
        go.Bar(
            x=df2["Date"],
            y=df2["Revenue (Billion USD)"],
            name="Revenue"
        ),
        row=2, col=1
    )

    fig.update_xaxes(title_text="Date", row=1, col=1)
    fig.update_xaxes(title_text="Year", row=2, col=1)
    fig.update_yaxes(title_text="Stock Price (USD)", row=1, col=1)
    fig.update_yaxes(title_text="Revenue (Billion USD)", row=2, col=1)

    fig.update_layout(
        title=f"{company} Stock Price and Revenue",
        height=900,
        width=1200,
        template="plotly_white"
    )

    fig.show()
```

```python
make_graph(df, df2, "Tesla")
```

```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots


# Prepare GameStop stock data
if "Date" not in df_2.columns:
    df_2 = df_2.rename_axis("Date").reset_index()

df_2["Date"] = pd.to_datetime(df_2["Date"])


# Prepare GameStop revenue data
if "Date" not in df_1.columns:
    df_1 = df_1.rename_axis("Date").reset_index()

df_1["Date"] = pd.to_datetime(df_1["Date"])

df_1 = df_1.dropna(subset=["Revenue (Billion USD)"])

df_1 = df_1.sort_values("Date")


# Define graph function
def make_graph(stock_data, revenue_data, company):

    fig = make_subplots(
        rows=2,
        cols=1,
        shared_xaxes=True,
        subplot_titles=(
            f"{company} Historical Share Price",
            f"{company} Historical Revenue"
        ),
        vertical_spacing=0.15
    )

    # Stock price graph
    fig.add_trace(
        go.Scatter(
            x=stock_data["Date"],
            y=stock_data["Close"],
            mode="lines",
            name="Stock Price"
        ),
        row=1,
        col=1
    )

    # Revenue graph
    fig.add_trace(
        go.Scatter(
            x=revenue_data["Date"],
            y=revenue_data["Revenue (Billion USD)"],
            mode="lines+markers",
            name="Revenue"
        ),
        row=2,
        col=1
    )

    # Axis labels
    fig.update_xaxes(title_text="Date", row=1, col=1)
    fig.update_xaxes(title_text="Year", row=2, col=1)

    fig.update_yaxes(title_text="Stock Price (USD)", row=1, col=1)
    fig.update_yaxes(title_text="Revenue (Billion USD)", row=2, col=1)

    # Layout
    fig.update_layout(
        title=f"{company} Stock Price and Revenue",
        title_x=0.5,
        height=1000,
        width=1200,
        showlegend=True,
        template="plotly_white"
    )

    fig.show()


make_graph(df_2, df_1, "GameStop")
```
